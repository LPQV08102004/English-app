```python
def create_filtered_csv(lang_filter='English', max_rows=None):
    rows = []
    
    print(f"Lọc các từ {lang_filter}...")
    
    with open(input_file, 'r', encoding='utf-8') as f:
        for i, line in enumerate(f):
            if max_rows and i >= max_rows:
                break
                
            try:
                data = json.loads(line.strip())
                
                # Chỉ lấy từ thuộc ngôn ngữ được chọn
                if data.get('lang') == lang_filter:
                    row = {
                        'word': data.get('word', ''),
                        'pos': data.get('pos', ''),
                        'etymology_text': data.get('etymology_text', ''),
                        'senses': json.dumps(data.get('senses', []), ensure_ascii=False),
                        'sounds': json.dumps(data.get('sounds', []), ensure_ascii=False),
                        'synonyms': json.dumps(data.get('synonyms', []), ensure_ascii=False),
                    }
                    rows.append(row)
                
                if (i + 1) % 50000 == 0:
                    print(f"Đã xử lý {i + 1:,} dòng, tìm thấy {len(rows):,} từ {lang_filter}...")
                    
            except json.JSONDecodeError:
                continue
    
    df = pd.DataFrame(rows)
    output_file = f"dictionary_{lang_filter.lower()}.csv"
    df.to_csv(output_file, index=False, encoding='utf-8-sig')
    
    print(f"\n✓ Hoàn thành!")
    print(f"  File: {output_file}")
    print(f"  Số từ {lang_filter}: {len(df):,}")
    
    return df

# Ví dụ: Xuất chỉ từ tiếng Anh
# Uncomment để chạy
# df_english = create_filtered_csv('English')
```

```python
# Xem chi tiết nội dung một số trường quan trọng
print("CHI TIẾT NỘI DUNG:")
print("=" * 80)

print(f"\n1. WORD: {entry.get('word', 'N/A')}")
print(f"2. LANGUAGE: {entry.get('lang', 'N/A')}")
print(f"3. POS (Part of Speech): {entry.get('pos', 'N/A')}")
print(f"4. LANG_CODE: {entry.get('lang_code', 'N/A')}")

print("\n5. ETYMOLOGY_TEXT:")
etymology = entry.get('etymology_text', '')
if etymology:
    print(f"   {etymology[:200]}...")
else:
    print("   (không có)")

print("\n6. SENSES (Nghĩa):")
senses = entry.get('senses', [])
print(f"   Số lượng nghĩa: {len(senses)}")
if senses:
    for i, sense in enumerate(senses[:6], 1):  # Chỉ hiển thị 6 nghĩa đầu
        glosses = sense.get('glosses', [])
        if glosses:
            print(f"   #{i}: {glosses[0]}")

print("\n7. SOUNDS (Phát âm):")
sounds = entry.get('sounds', [])
print(f"   Số lượng phát âm: {len(sounds)}")
if sounds:
    for sound in sounds[:2]:
        ipa = sound.get('ipa', '')
        if ipa:
            print(f"   - IPA: {ipa}")

print("\n8. FORMS (Dạng từ):")
forms = entry.get('forms', [])
print(f"   Số lượng dạng: {len(forms)}")
if forms:
    for form in forms[:3]:
        form_name = form.get('form', '')
        tags = form.get('tags', [])
        print(f"   - {form_name}: {', '.join(tags)}")

print("\n9. SYNONYMS (Từ đồng nghĩa):")
synonyms = entry.get('synonyms', [])
print(f"   Số lượng: {len(synonyms)}")
if synonyms:
    for syn in synonyms[:3]:
        word = syn.get('word', '')
        if word:
            print(f"   - {word}")

print("\n10. TRANSLATIONS (Bản dịch):")
translations = entry.get('translations', [])
print(f"   Số lượng: {len(translations)}")
if translations:
    for trans in translations[:550]:
        lang = trans.get('lang', '')
        word = trans.get('word', '')
        lang_code = trans.get('code', '')
        if word:
            print(f"   - {lang} ({lang_code}): {word}")

print("\n" + "=" * 80)
```

## Chuyển đổi Excel sang JSON

Chuyển file Excel "Bảng Từ Vựng Chủ Đề _School_.xlsx" sang định dạng JSON:

```python
import pandas as pd
import json

# Đọc file Excel
excel_file = r"Bảng Từ Vựng Chủ Đề _School_.xlsx"
df = pd.read_excel(excel_file)

print("=" * 80)
print("THÔNG TIN FILE EXCEL")
print("=" * 80)
print(f"Số dòng: {len(df)}")
print(f"Số cột: {len(df.columns)}")
print(f"\nTên các cột:")
for i, col in enumerate(df.columns, 1):
    print(f"  {i}. {col}")

print("\n" + "=" * 80)
print("XEM DỮ LIỆU MẪU (5 dòng đầu):")
print("=" * 80)
print(df.head())

# Chuyển đổi sang JSON
output_json = "vocabulary_school.json"

# Option 1: JSON dạng mảng các object (phổ biến nhất)
df.to_json(output_json, orient='records', force_ascii=False, indent=2)
print(f"\n✓ Đã lưu JSON dạng records: {output_json}")

# Option 2: JSON dạng khác (nếu cần)
# df.to_json("vocabulary_school_table.json", orient='table', force_ascii=False, indent=2)
# print(f"✓ Đã lưu JSON dạng table: vocabulary_school_table.json")

# Xem nội dung JSON vừa tạo
print("\n" + "=" * 80)
print("NỘI DUNG JSON (3 mục đầu tiên):")
print("=" * 80)
with open(output_json, 'r', encoding='utf-8') as f:
    data = json.load(f)
    print(json.dumps(data[:3], ensure_ascii=False, indent=2))

print(f"\n✓ Hoàn thành! Tổng số mục trong JSON: {len(data)}")
```
